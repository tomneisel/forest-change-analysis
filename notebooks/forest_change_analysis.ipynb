```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import rasterio
import joblib
```

```python
path_B01_2018 = "../data/raw/satellite/merged/2018/2018_B01_harz.tif"
path_B02_2018 = "../data/raw/satellite/merged/2018/2018_B02_harz.tif"
path_B03_2018 = "../data/raw/satellite/merged/2018/2018_B03_harz.tif"
path_B04_2018 = "../data/raw/satellite/merged/2018/2018_B04_harz.tif"
path_B05_2018 = "../data/raw/satellite/merged/2018/2018_B05_harz.tif"
path_B06_2018 = "../data/raw/satellite/merged/2018/2018_B06_harz.tif"
path_B07_2018 = "../data/raw/satellite/merged/2018/2018_B07_harz.tif"
path_B8A_2018 = "../data/raw/satellite/merged/2018/2018_B08_harz.tif"
path_B11_2018 = "../data/raw/satellite/merged/2018/2018_B11_harz.tif"
path_B12_2018 = "../data/raw/satellite/merged/2018/2018_B12_harz.tif"

path_B01_2026 = "../data/raw/satellite/merged/2026/2026_B01_harz.tif"
path_B02_2026 = "../data/raw/satellite/merged/2026/2026_B02_harz.tif"
path_B03_2026 = "../data/raw/satellite/merged/2026/2026_B03_harz.tif"
path_B04_2026 = "../data/raw/satellite/merged/2026/2026_B04_harz.tif"
path_B05_2026 = "../data/raw/satellite/merged/2026/2026_B05_harz.tif"
path_B06_2026 = "../data/raw/satellite/merged/2026/2026_B06_harz.tif"
path_B07_2026 = "../data/raw/satellite/merged/2026/2026_B07_harz.tif"
path_B8A_2026 = "../data/raw/satellite/merged/2026/2026_B08_harz.tif"
path_B11_2026 = "../data/raw/satellite/merged/2026/2026_B11_harz.tif"
path_B12_2026 = "../data/raw/satellite/merged/2026/2026_B12_harz.tif"

def get_image_data(path):

    src = rasterio.open(path)
    data = src.read(1)
    return data, src

data_B01_2018, src_B01_2018 = get_image_data(path_B01_2018)
data_B02_2018, src_B02_2018 = get_image_data(path_B02_2018)
data_B03_2018, src_B03_2018 = get_image_data(path_B03_2018)
data_B04_2018, src_B04_2018 = get_image_data(path_B04_2018)
data_B05_2018, src_B05_2018 = get_image_data(path_B05_2018)
data_B06_2018, src_B06_2018 = get_image_data(path_B06_2018)
data_B07_2018, src_B07_2018 = get_image_data(path_B07_2018)
data_B8A_2018, src_B8A_2018 = get_image_data(path_B8A_2018)
data_B11_2018, src_B11_2018 = get_image_data(path_B11_2018)
data_B12_2018, src_B12_2018 = get_image_data(path_B12_2018)

data_B01_2026, src_B01_2026 = get_image_data(path_B01_2026)
data_B02_2026, src_B02_2026 = get_image_data(path_B02_2026)
data_B03_2026, src_B03_2026 = get_image_data(path_B03_2026)
data_B04_2026, src_B04_2026 = get_image_data(path_B04_2026)
data_B05_2026, src_B05_2026 = get_image_data(path_B05_2026)
data_B06_2026, src_B06_2026 = get_image_data(path_B06_2026)
data_B07_2026, src_B07_2026 = get_image_data(path_B07_2026)
data_B8A_2026, src_B8A_2026 = get_image_data(path_B8A_2026)
data_B11_2026, src_B11_2026 = get_image_data(path_B11_2026)
data_B12_2026, src_B12_2026 = get_image_data(path_B12_2026)

#reshaping (1306, 1400) to one dimension
data_B01_2018 = data_B01_2018.reshape(-1,)
data_B02_2018 = data_B02_2018.reshape(-1,)
data_B03_2018 = data_B03_2018.reshape(-1,)
data_B04_2018 = data_B04_2018.reshape(-1,)
data_B05_2018 = data_B05_2018.reshape(-1,)
data_B06_2018 = data_B06_2018.reshape(-1,)
data_B07_2018 = data_B07_2018.reshape(-1,)
data_B8A_2018 = data_B8A_2018.reshape(-1,)
data_B11_2018 = data_B11_2018.reshape(-1,)
data_B12_2018 = data_B12_2018.reshape(-1,)

data_B01_2026 = data_B01_2026.reshape(-1,)
data_B02_2026 = data_B02_2026.reshape(-1,)
data_B03_2026 = data_B03_2026.reshape(-1,)
data_B04_2026 = data_B04_2026.reshape(-1,)
data_B05_2026 = data_B05_2026.reshape(-1,)
data_B06_2026 = data_B06_2026.reshape(-1,)
data_B07_2026 = data_B07_2026.reshape(-1,)
data_B8A_2026 = data_B8A_2026.reshape(-1,)
data_B11_2026 = data_B11_2026.reshape(-1,)
data_B12_2026 = data_B12_2026.reshape(-1,)
```

```python
data_2018 = {"B01": data_B01_2018, "B02": data_B02_2018, "B03": data_B03_2018, "B04":data_B04_2018, "B05": data_B05_2018, "B06": data_B06_2018, "B07": data_B07_2018, "B8A":data_B8A_2018, "B11": data_B11_2018, "B12": data_B12_2018}
satellite_harz_2018_df = pd.DataFrame(data_2018)
satellite_harz_2018_df
```

```python
data_2026 = {"B01": data_B01_2026, "B02": data_B02_2026, "B03": data_B03_2026, "B04":data_B04_2026, "B05": data_B05_2026, "B06": data_B06_2026, "B07": data_B07_2026, "B8A":data_B8A_2026, "B11": data_B11_2026, "B12": data_B12_2026}
satellite_harz_2026_df = pd.DataFrame(data_2026)
satellite_harz_2026_df
```

```python
clf_forest = joblib.load("../models/forest_classifier.joblib")
```

```python
valid_mask = (satellite_harz_2018_df != 0).all(axis=1)

valid_2018_df = satellite_harz_2018_df.loc[valid_mask]
valid_2026_df = satellite_harz_2026_df.loc[valid_mask]

predictions_2018 = clf_forest.predict(valid_2018_df)
predictions_2026 = clf_forest.predict(valid_2026_df)

forest_2018_target = np.full(len(satellite_harz_2018_df), 2, dtype="uint8")
forest_2018_target[valid_mask] = predictions_2018
forest_2026_target = np.full(len(satellite_harz_2026_df), 2, dtype="uint8")
forest_2026_target[valid_mask] = predictions_2026

forest_2018_target

satellite_harz_2018_df["target"] = forest_2018_target
satellite_harz_2026_df["target"] = forest_2026_target

```

```python
satellite_harz_2018_df[satellite_harz_2018_df == 2] = np.nan
satellite_harz_2026_df[satellite_harz_2026_df == 2] = np.nan
```

```python
satellite_harz_2026_df["target"].value_counts()
```

```python
#comparing the 3 years
fig, axes = plt.subplots(1, 2, figsize = (12, 6))

axes[0].imshow(np.array(satellite_harz_2018_df["target"]).reshape(1306, 1400), cmap="Greens")
axes[1].imshow(np.array(satellite_harz_2026_df["target"]).reshape(1306, 1400), cmap="Greens")

axes[0].axis("off")
axes[1].axis("off")

axes[0].set_title("2018")
axes[1].set_title("2026")

fig.suptitle("forest prediction of the model for the years 2018 and 2026")
```
