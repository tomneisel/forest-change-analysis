```python
data_2018 = {"B01": data_B01_2018, "B02": data_B02_2018, "B03": data_B03_2018, "B04":data_B04_2018, "B05": data_B05_2018, "B06": data_B06_2018, "B07": data_B07_2018, "B8A":data_B8A_2018, "B11": data_B11_2018, "B12": data_B12_2018}
satellite_harz_2018_df = pd.DataFrame(data_2018)
satellite_harz_2018_df
```

```python
data_2026 = {"B01": data_B01_2026, "B02": data_B02_2026, "B03": data_B03_2026, "B04":data_B04_2026, "B05": data_B05_2026, "B06": data_B06_2026, "B07": data_B07_2026, "B8A":data_B8A_2026, "B11": data_B11_2026, "B12": data_B12_2026}
satellite_harz_2026_df = pd.DataFrame(data_2026)
satellite_harz_2026_df
```

```python
clf_forest = joblib.load("../models/forest_classifier.joblib")
```

```python
valid_mask = (satellite_harz_2018_df != 0).all(axis=1)

valid_2018_df = satellite_harz_2018_df.loc[valid_mask]
valid_2026_df = satellite_harz_2026_df.loc[valid_mask]

predictions_2018 = clf_forest.predict(valid_2018_df)
predictions_2026 = clf_forest.predict(valid_2026_df)

forest_2018_target = np.full(len(satellite_harz_2018_df), 2, dtype="uint8")
forest_2018_target[valid_mask] = predictions_2018
forest_2026_target = np.full(len(satellite_harz_2026_df), 2, dtype="uint8")
forest_2026_target[valid_mask] = predictions_2026

forest_2018_target

satellite_harz_2018_df["target"] = forest_2018_target
satellite_harz_2026_df["target"] = forest_2026_target

```

```python
satellite_harz_2018_df[satellite_harz_2018_df == 2] = np.nan
satellite_harz_2026_df[satellite_harz_2026_df == 2] = np.nan

harz_2018_data = satellite_harz_2018_df["target"]
harz_2026_data = satellite_harz_2026_df["target"]

harz_2018_image = np.array(harz_2018_data).reshape(1306, 1400)
harz_2026_image = np.array(harz_2026_data).reshape(1306, 1400)
```

```python
#comparing the 3 years
fig, axes = plt.subplots(1, 2, figsize = (12, 6))

axes[0].imshow(harz_2018_image, cmap="Greens")
axes[1].imshow(harz_2026_image, cmap="Greens")

axes[0].axis("off")
axes[1].axis("off")

axes[0].set_title("2018")
axes[1].set_title("2026")

fig.suptitle("forest prediction of the model for the years 2018 and 2026")
```

```python
forest_decrease = 100 * (harz_2018_data[harz_2018_data == 1].sum() - harz_2026_data[harz_2026_data == 1].sum()) / harz_2018_data[harz_2018_data == 1].sum()
print(f"predicted forest decrease of {round(forest_decrease, 1)}%")
```

```python
change_map = harz_2026_image - harz_2018_image

fig, ax = plt.subplots(figsize = (8, 8))

ax.imshow(change_map, cmap="RdYlGn")

legend_elements = [
    Patch(facecolor="red", label="Forest loss"),
    Patch(facecolor="yellow", label="Stable"),
    Patch(facecolor="green", label="Forest gain")
]

ax.set_title("Development of Forest Area from 2018 to 2026")
ax.legend(handles=legend_elements, loc="upper left")
ax.axis("off")
```

```python
window_size = 27
area_is_valid = ~np.isnan(change_map)
values = np.nan_to_num(change_map, nan=0)

local_sum = uniform_filter(
    values.astype(float),
    size = window_size
) * window_size ** 2

local_count = uniform_filter(
    area_is_valid.astype(float),
    size=window_size
) * window_size**2

local_mean = local_sum / local_count
local_mean[~area_is_valid] = np.nan


fig, ax = plt.subplots(figsize = (10, 8))

im = ax.imshow(local_mean.reshape(1306, 1400), cmap="RdYlGn", vmin=-1, vmax=1)
cbar = fig.colorbar(im, ax=ax)
cbar.set_ticks([-1, 0, 1])
cbar.set_ticklabels([
    "Forest loss",
    "Stable",
    "Forest gain"
])
ax.set_title("Local Forest Change Intensity")
ax.axis("off")
```
